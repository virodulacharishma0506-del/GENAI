```python
%%writefile budget_estimator.py
from dataclasses import dataclass

@dataclass
class BudgetReport:
    monthly_requests: float
    monthly_input_tokens: float
    monthly_output_tokens: float
    monthly_total_tokens: float
    total_cost_usd: float
    total_cost_inr: float
    peak_rpm: float
    peak_tpm: float
    within_budget: bool
    within_rpm: bool
    within_tpm: bool


def calculate_budget(
    daily_active_users,
    requests_per_user_per_day,
    avg_input_tokens,
    avg_output_tokens,
    in_price_usd_per_1m,
    out_price_usd_per_1m,
    usd_to_inr,
    provider_rpm_limit,
    provider_tpm_limit,
    monthly_budget_inr,
    days_per_month=30,
    peak_traffic_multiplier=3.0
):
    daily_requests = daily_active_users * requests_per_user_per_day
    monthly_requests = daily_requests * days_per_month

    input_tokens = monthly_requests * avg_input_tokens
    output_tokens = monthly_requests * avg_output_tokens
    total_tokens = input_tokens + output_tokens

    input_cost = (input_tokens / 1_000_000) * in_price_usd_per_1m
    output_cost = (output_tokens / 1_000_000) * out_price_usd_per_1m

    total_usd = input_cost + output_cost
    total_inr = total_usd * usd_to_inr

    avg_rpm = daily_requests / 1440
    peak_rpm = avg_rpm * peak_traffic_multiplier

    peak_tpm = peak_rpm * (avg_input_tokens + avg_output_tokens)

    return BudgetReport(
        monthly_requests,
        input_tokens,
        output_tokens,
        total_tokens,
        total_usd,
        total_inr,
        peak_rpm,
        peak_tpm,
        total_inr <= monthly_budget_inr,
        peak_rpm <= provider_rpm_limit,
        peak_tpm <= provider_tpm_limit
    )
```

```python
%%writefile test_budget.py

import unittest
from budget_estimator import calculate_budget


class TestLLMBudgetEstimator(unittest.TestCase):

    def setUp(self):
        self.base_params = {
            "daily_active_users": 1000,
            "requests_per_user_per_day": 2,
            "avg_input_tokens": 1000,
            "avg_output_tokens": 500,
            "in_price_usd_per_1m": 1.00,
            "out_price_usd_per_1m": 4.00,
            "usd_to_inr": 90.0,
            "provider_rpm_limit": 100,
            "provider_tpm_limit": 200000,
            "monthly_budget_inr": 20000.0,
            "days_per_month": 30,
            "peak_traffic_multiplier": 3.0,
        }

    def test_token_and_inr_cost_calculation(self):
        report = calculate_budget(**self.base_params)

        self.assertEqual(report.monthly_requests, 60000)
        self.assertEqual(report.monthly_input_tokens, 60000000)
        self.assertEqual(report.monthly_output_tokens, 30000000)
        self.assertAlmostEqual(report.total_cost_usd, 180.00)
        self.assertAlmostEqual(report.total_cost_inr, 16200.00)
        self.assertTrue(report.within_budget)

    def test_budget_exceeded(self):
        params = {
            **self.base_params,
            "monthly_budget_inr": 15000.0
        }
        report = calculate_budget(**params)

        self.assertFalse(report.within_budget)

    def test_rate_limits_within_capacity(self):
        report = calculate_budget(**self.base_params)

        self.assertAlmostEqual(report.peak_rpm, 4.1667, places=3)
        self.assertAlmostEqual(report.peak_tpm, 6250.0, places=1)
        self.assertTrue(report.within_rpm)
        self.assertTrue(report.within_tpm)

    def test_rpm_and_tpm_limits_exceeded(self):
        params = {
            **self.base_params,
            "daily_active_users": 50000,
            "provider_rpm_limit": 200,
            "provider_tpm_limit": 250000,
        }
        report = calculate_budget(**params)

        self.assertFalse(report.within_rpm)
        self.assertFalse(report.within_tpm)

    def test_zero_traffic_edge_case(self):
        params = {
            **self.base_params,
            "daily_active_users": 0
        }
        report = calculate_budget(**params)

        self.assertEqual(report.monthly_requests, 0)
        self.assertEqual(report.total_cost_inr, 0.0)
        self.assertTrue(report.within_budget)
        self.assertTrue(report.within_rpm)
        self.assertTrue(report.within_tpm)


if __name__ == "__main__":
    unittest.main()
```
